KNN - K Nearest Neighbours is a classificaiton/regression algorithm that assigns data points specific classes by their position in vector space. It is calculated by the Euclidean formula which is just pithagora.

For a new data point x and all the data points in the dataset

the Euclidean distance is $d(x, y) = \sqrt{(x_1 - y_1)^2 + (x_2 - y_2)^2 + \cdots + (x_n - y_n)^2}$. 

Additionally we specify k which is points closest to x

#### 

```python
# Here you can see a simple approach using a predifined classifier in sklearn 
from sklearn.neighbors import KNeighborsClassifier
import numpy as np
# sample data 
X = np.array([[1, 1], [2, 2], [1.5, 1.5]])
y = np.array(['A', 'B', 'A'])

knn = KNeighborsClassifier(n_neighbors=2)
knn.fit(X, y)

new_point = np.array([[1.2, 1.3]])
prediction = knn.predict(new_point)

print(f'Estimate class is {prediction}')
```

```python
from collections import Counter
# Let's try to implement knn on our own which is not so obvious
X = np.array([[1, 1], [2, 2], [1.5, 1.5]])
y = np.array(['A', 'B', 'A'])
datapoint = np.array([[1.2, 1.3]])

def euclidean_distance(x1, x2):
    return np.sqrt(np.sum((x1 - x2) ** 2))

def knnclassifier(X_train, y_train, x_query, k=3):
    distance = np.array([euclidean_distance(X_train, x) for x in x_query])
    k_indices = np.argsort(distance)[:k] # returns the indices of the lowest values until k 
    prediction = Counter(np.array(y_train)[k_indices]).most_common(1)[0][0]
    return prediction


knn = knnclassifier(X, y, datapoint)
print(knn)
    
```
