```python
import numpy as np
```

Standardization(Z-score) is a process of scaling features so they have mean = 0 and standard deviation of 1
The formula is z = x - μ​ / σ

Where:

x is the feature

μ is the feature's mean 

σ is the feature's standard deviation

```python
A = [2, 4, 5, 7, 9]
mean = np.mean(A)
standard_deviation = np.std(A)
print(f"Mean: {mean}")
print(f"Standard Deviation: {standard_deviation}")
# mean/μ =  sum(xi) / N |WHERE| N - len(A)
# std/σ = xi - μi 
z = (A - mean) / standard_deviation
print(z)
```

Min-Max scaling(Normalization)

Min-Max is used to scale features into a fixed shpae [0, 1] or [-1, 1]

x' = (X - Xmin) / (Xmax - Xmin)

```python
max_val = np.max(A)
min_val = np.min(A)

x = (A - min_val) / (max_val - min_val)
print(x) 
```
