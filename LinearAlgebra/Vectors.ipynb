```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
# Length of a vector:

A = np.array([1, 2])
lengthA = np.linalg.norm(A)
print(f'Length of vector A: ', lengthA)
```

```python
# Projections

A = np.array([3, 4])
B = np.array([2, 0])

AprojB = ((np.dot(A, B) / np.dot(B, B)) * B)

print(f'Projection of A onto B is: ', AprojB)
```

```python
# Changing basis of a vector in a vector space

# Standart vector space with i=(1, 0) j=(0, 1)


basis = np.array([[1, -1],
                  [1, 1]])

vector = np.array([2, 3])

inverse_basis = np.linalg.inv(basis)
vect_in_new_basis = np.dot(inverse_basis, vector)
print(f'Inverse basis: ', inverse_basis)
print(f'Vector in new inverted basis:\n', vect_in_new_basis)
```

### Normalizations
L1, L2, Linfinity

```python
# Just some vectors
vector = np.array([
    [3, 2, 2],
])

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.quiver(0, 0, 0, vector[0, 0], vector[0, 1], vector[0, 2], color='r', arrow_length_ratio=0.1, label='v1')

ax.set_xlim(0, 5)
ax.set_ylim(0, 5)
ax.set_zlim(0, 5)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.legend()
plt.show()
```

```python
def vector_normalization(vector):
    l1_norm = np.linalg.norm(vector, 1)
    l2_norm = np.linalg.norm(vector, 2)
    linf_norm = np.linalg.norm(vector, np.inf)

    L1 = vector / l1_norm
    L2 = vector / l2_norm
    Linf = vector / linf_norm

    return (L1, L2, Linf, l1_norm, l2_norm, linf_norm)


x_l1, x_l2, x_linf, l1, l2, linf = vector_normalization(vector)

print("L1 norm:", l1, "\nL1 normalized:\n", x_l1)
print("L2 norm:", l2, "\nL2 normalized:\n", x_l2)
print("L∞ norm:", linf, "\nL∞ normalized:\n", x_linf)
```

```python
L1, L2, Linf, l1_norm, l2_norm, linf_norm = vector_normalization(vector)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.quiver(0, 0, 0, L1[0, 0], L1[0, 1], L1[0, 2], color='r', arrow_length_ratio=0.1, label='L1 norm')
ax.quiver(0, 0, 0, L2[0, 0], L2[0, 1], L2[0, 2], color='b', arrow_length_ratio=0.1, label='L2 norm')
ax.quiver(0, 0, 0, Linf[0, 0], Linf[0, 1], Linf[0, 2], color='g', arrow_length_ratio=0.1, label='L∞ norm')

ax.set_xlim(0, 5)
ax.set_ylim(0, 5)
ax.set_zlim(0, 5)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.legend()
plt.show()
```

#### What happens if we multiply a norm by a scalar? 

#### ∥λv∥=∣λ∣⋅∥v∥

```python
vector = np.array([3, 4])
scalar = 2

l1_norm = np.linalg.norm(vector, 1) 
l2_norm = np.linalg.norm(vector, 2) 

scaled_l1_norm = scalar * l1_norm
scaled_l2_norm = scalar * l2_norm

L1_normed = vector / l1_norm
L2_normed = vector / l2_norm

scaled_L1_normed = vector / scaled_l1_norm
scaled_L2_normed = vector / scaled_l2_norm

print(f"L1 norm: {l1_norm}")
print(f"Scaled L1 norm: {scaled_l1_norm}")
print(f"L1 normalized: {L1_normed}")
print(f"Scaled L1 normalized: {scaled_L1_normed}\n")

print(f"L2 norm: {l2_norm}")
print(f"Scaled L2 norm: {scaled_l2_norm}")
print(f"L2 normalized: {L2_normed}")
print(f"Scaled L2 normalized: {scaled_L2_normed}")
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.quiver(0, 0, L1_normed[0], L1_normed[1], angles='xy', scale_units='xy', scale=1, color='r', width=0.006, label='L1 normalized')
ax1.quiver(0, 0, scaled_L1_normed[0], scaled_L1_normed[1], angles='xy', scale_units='xy', scale=1, color='darkred', width=0.006, label='Scaled L1 normalized')
ax1.set_xlim(0, 1)
ax1.set_ylim(0, 1)
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_title('L1 Normalization')
ax1.grid(True)
ax1.legend()
ax1.set_aspect('equal')

ax2.quiver(0, 0, L2_normed[0], L2_normed[1], angles='xy', scale_units='xy', scale=1, color='b', width=0.006, label='L2 normalized')
ax2.quiver(0, 0, scaled_L2_normed[0], scaled_L2_normed[1], angles='xy', scale_units='xy', scale=1, color='darkblue', width=0.006, label='Scaled L2 normalized')
ax2.set_xlim(0, 1)
ax2.set_ylim(0, 1)
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_title('L2 Normalization')
ax2.grid(True)
ax2.legend()
ax2.set_aspect('equal')

plt.tight_layout()
plt.show()
```

So we can see that the normalized vector which normalization was multiplied by 2 is twice as the original one's length, so it's been shrinked.
