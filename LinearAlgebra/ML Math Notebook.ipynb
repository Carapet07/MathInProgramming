```python
import numpy as np
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Inverse of a matrix
A = np.array([[1, 2],
              [3, 4]], dtype=float)

A_inv = np.linalg.inv(A)

print(A)
print(A_inv)
matrix_multiplication = A @ A_inv
print(f'A @ A_inv:\n', np.round(matrix_multiplication, 3))
```

```python
# Try to compute inverse of a matrix which det = 0

A = np.array([[2, 4],
              [1, 2]], dtype=float)

A_det = np.linalg.det(A)
print('Determinant of A: ', A_det)
try:
  A_inv = np.linalg.inv(A)

except np.linalg.LinAlgError as e:
  print(f'Matrix is not invertible', e)

```

```python
# Check if the transposed inverse of a matrix == inversed transpose of a matrix:
# (AT)^−1=(A−1)^T

A = np.array([[3, 4],
             [1, 2]], dtype=float)


lhs = np.linalg.inv(A.T)
rhs = np.linalg.inv(A).T
print(lhs)
print(rhs)
print(np.allclose(lhs, rhs))
```

```python
# Length of a vector:

A = np.array([1, 2])
lengthA = np.linalg.norm(A)
print(f'Length of vector A: ', lengthA)
```

```python
# Projections

A = np.array([3, 4])
B = np.array([2, 0])

AprojB = ((np.dot(A, B) / np.dot(B, B)) * B)

print(f'Projection of A onto B is: ', AprojB)
```

```python
# Changing basis of a vector in a vector space

# Standart vector space with i=(1, 0) j=(0, 1)


basis = np.array([[1, -1],
                  [1, 1]])

vector = np.array([2, 3])

inverse_basis = np.linalg.inv(basis)
vect_in_new_basis = np.dot(inverse_basis, vector)
print(f'Inverse basis: ', inverse_basis)
print(f'Vector in new inverted basis:\n', vect_in_new_basis)

```

### Normalizations
L1, L2, Linfinity

```python
# Just some vecotrs
vector = np.array([
    [3, 2],
    [2, 7]
])

plt.figure()
plt.quiver(0, 0, vector[0, 0], vector[0, 1], angles='xy', scale_units='xy', scale=1, color='r', label='v1')
plt.quiver(0, 0, vector[1, 0], vector[1, 1], angles='xy', scale_units='xy', scale=1, color='b', label='v2')

plt.xlim(0, 8)
plt.ylim(0, 8)
plt.grid(True)
plt.legend()
plt.gca().set_aspect('equal')
plt.show()

```

```python
def vector_normalization(vector):
    # Compute norms
    l1_norm = np.linalg.norm(vector, 1, axis=1, keepdims=True)
    l2_norm = np.linalg.norm(vector, 2, axis=1, keepdims=True)
    linf_norm = np.linalg.norm(vector, np.inf, axis=1, keepdims=True)

    # Normalized vectors
    L1 = vector / l1_norm
    L2 = vector / l2_norm
    Linf = vector / linf_norm

    return (L1, L2, Linf, l1_norm, l2_norm, linf_norm)


x_l1, x_l2, x_linf, l1, l2, linf = vector_normalization(vector)

print("L1 norm:", l1, "\nL1 normalized:\n", x_l1)
print("L2 norm:", l2, "\nL2 normalized:\n", x_l2)
print("L∞ norm:", linf, "\nL∞ normalized:\n", x_linf)
```

```python
normalised_vectors = vector_normalization(vector)

plt.figure()
plt.quiver(0, 0, normalised_vectors[0, 0], normalised_vectors[0, 1], angles='xy', scale_units='xy', scale=1, color='r', label='v1')
plt.quiver(0, 0, normalised_vectors[1, 0], normalised_vectors[1, 1], angles='xy', scale_units='xy', scale=1, color='b', label='v2')

plt.xlim(0, 8)
plt.ylim(0, 8)
plt.grid(True)
plt.legend()
plt.gca().set_aspect('equal')
plt.show()
```
